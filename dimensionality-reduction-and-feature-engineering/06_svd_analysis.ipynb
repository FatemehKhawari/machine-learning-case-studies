```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import load_breast_cancer
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import TruncatedSVD
```

```python
# Load Dataset
data = load_breast_cancer()
x = pd.DataFrame (data.data , columns = data.feature_names)
y = pd.Series(data.target)
x.head()
```

```python
#Standardized data
scaler = StandardScaler()
x_scaled = scaler.fit_transform(x)
```

```python
#Apply SVD
#Reduce features into 2 dimensions/ 569 samples × 30 features TO 569 samples × 2 components
svd = TruncatedSVD(n_components = 2 , random_state = 42)
x_svd = svd.fit_transform(x_scaled)
```

```python
#Create DataFrame
svd_df = pd.DataFrame(x_svd , columns = ["Component 1" , "Component 2"])
svd_df["target"] = y.values
svd_df.head()
```

```python
# explained variance ration
print(
    svd.explained_variance_ratio_
)
#Component 1 explains 44% of variation
#Component 2 explains 18%
```

```python
#Total
total_variance = sum(svd.explained_variance_ratio_)
print (total_variance)
```

```python
# Visualize SVD projection

plt.figure(figsize = (8,6))
sns.scatterplot(
    data = svd_df,
    x = "Component 1",
    y = "Component 2",
    hue = "target" ,
    palette = "viridis"
)

plt.title(" SVD Projection of Breast Cancer")

plt.show()
```

```python
#Singular Values
# A larger singular value means:more information and more contribution to reconstruction

print(
    svd.singular_values_
)
```

```python
# Visualize singular values
plt.figure(figsize = (5,8))
plt.plot(
    range(1 , len(svd.singular_values_)+1 ),
    svd.singular_values_,
    marker = "o"
)

plt.xlabel("Component")
plt.ylabel(" Singular Value")
plt.title(" SVD Singular Values")

plt.grid()
plt.show()
```

```python
# SVD Components (feature contribution)

components = pd.DataFrame(
    svd.components_,
    columns=x.columns,
    index=[
        "Component 1",
        "Component 2"
    ]
)

components
```
